1. Create a 1D NumPy array of integers from 10 to 50 (inclusive, step 5). Print the array, its shape, dtype and number of dimensions.

```python
arr1 = np.arange(10, 51, 5)

print("Array:", arr1)
print("Shape:", arr1.shape)
print("Data type:", arr1.dtype)
print("Number of dimensions:", arr1.ndim)
```

2. Create a 3x4 array of zeros, a 3x4 array of ones, and a 4x4 identity matrix. Then create an array of 8 evenly spaced numbers between 0 and 1 using linspace.

```python
zeros_arr = np.zeros((3, 4))
ones_arr = np.ones((3, 4))
identity_arr = np.eye(4)
lin_arr = np.linspace(0, 1, 8)

print("Zeros array:\n", zeros_arr)
print("Ones array:\n", ones_arr)
print("Identity matrix:\n", identity_arr)
print("Linspace array:", lin_arr)
```

3. Create a 1D array of numbers from 1 to 24 and reshape it into a 2D array of shape (4, 6). Then reshape it into a 3D array of shape (2, 3, 4) and print the shape of each.

```python
arr3 = np.arange(1, 25)
print("1D array:", arr3)
print("1D shape:", arr3.shape)

arr_2d = arr3.reshape(4, 6)
print("2D array:\n", arr_2d)
print("2D shape:", arr_2d.shape)

arr_3d = arr3.reshape(2, 3, 4)
print("3D array:\n", arr_3d)
print("3D shape:", arr_3d.shape)
```

4. Given arr = np.array([12, 45, 7, 23, 56, 89, 34, 21]), print: the first 3 elements, the last 3 elements, every alternate element, and the array in reverse order.

```python
arr4 = np.array([12, 45, 7, 23, 56, 89, 34, 21])

print("First 3 elements:", arr4[:3])
print("Last 3 elements:", arr4[-3:])
print("Every alternate element:", arr4[::2])
print("Reverse order:", arr4[::-1])
```

5. Create a 2D array of shape (5, 5) with values 1 to 25. Extract (a) the second row, (b) the third column, (c) the 3x3 sub-matrix from the center.

```python
arr5 = np.arange(1, 26).reshape(5, 5)
print("Full array:\n", arr5)

print("(a) Second row:", arr5[1])
print("(b) Third column:", arr5[:, 2])
print("(c) Center 3x3 sub-matrix:\n", arr5[1:4, 1:4])
```

6. Given a = np.array([10, 20, 30, 40]) and b = np.array([1, 2, 3, 4]), perform addition, subtraction, multiplication, division and power (a ** 2). Also compute the dot product of a and b.

```python
a = np.array([10, 20, 30, 40])
b = np.array([1, 2, 3, 4])

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Power (a ** 2):", a ** 2)
print("Dot product:", np.dot(a, b))
```

7. Given arr = np.array([[4, 8, 2], [9, 1, 7], [5, 6, 3]]), find the sum, mean, min, max and standard deviation of the whole array, and then the sum along rows (axis=1) and along columns (axis=0).

```python
arr7 = np.array([[4, 8, 2],
                 [9, 1, 7],
                 [5, 6, 3]])

print("Sum:", np.sum(arr7))
print("Mean:", np.mean(arr7))
print("Min:", np.min(arr7))
print("Max:", np.max(arr7))
print("Standard deviation:", np.std(arr7))

print("Sum along rows (axis=1):", np.sum(arr7, axis=1))
print("Sum along columns (axis=0):", np.sum(arr7, axis=0))
```

8. Create an array of 20 random integers between 1 and 100. Using boolean masking, print (a) all elements greater than 50, (b) all even elements, (c) count of elements divisible by 5.

```python
arr8 = np.random.randint(1, 101, 20)
print("Random array:", arr8)

# (a) elements greater than 50
print("(a) Greater than 50:", arr8[arr8 > 50])

# (b) even elements
print("(b) Even elements:", arr8[arr8 % 2 == 0])

# (c) count of elements divisible by 5
count = np.sum(arr8 % 5 == 0)
print("(c) Count divisible by 5:", count)
```

9. Use np.where to replace all negative values in np.array([5, -3, 8, -1, 0, -7, 4]) with 0. Also find the index of the maximum and minimum value using argmax and argmin.

```python
arr9 = np.array([5, -3, 8, -1, 0, -7, 4])

new_arr = np.where(arr9 < 0, 0, arr9)
print("Original array:", arr9)
print("After replacing negatives with 0:", new_arr)

print("Index of maximum value:", np.argmax(arr9))
print("Index of minimum value:", np.argmin(arr9))
```

10. Create two 2x3 arrays and join them vertically (vstack) and horizontally (hstack). Then sort the array np.array([[9, 3, 6], [2, 8, 1]]) along each row and flatten the result into 1D.

```python
x = np.array([[1, 2, 3],
              [4, 5, 6]])
y = np.array([[7, 8, 9],
              [10, 11, 12]])

print("Vertical stack:\n", np.vstack((x, y)))
print("Horizontal stack:\n", np.hstack((x, y)))

arr10 = np.array([[9, 3, 6],
                  [2, 8, 1]])

sorted_arr = np.sort(arr10, axis=1)
print("Sorted along each row:\n", sorted_arr)

flat_arr = sorted_arr.flatten()
print("Flattened 1D array:", flat_arr)
```
