1. Create a DataFrame from a dictionary with columns Name, Age, City and Salary for 6 employees. Print the first 3 rows, the shape, the column names and the data types.

```python
data = {
    "Name": ["Ali", "Sara", "Ahmed", "Zara", "Usman", "Hina"],
    "Age": [25, 32, 28, 35, 41, 30],
    "City": ["Delhi", "Mumbai", "Delhi", "Pune", "Delhi", "Mumbai"],
    "Salary": [35000, 52000, 45000, 60000, 48000, 41000]
}

df = pd.DataFrame(data)

print("First 3 rows:\n", df.head(3))
print("Shape:", df.shape)
print("Column names:", list(df.columns))
print("Data types:\n", df.dtypes)
```

6. Add a new column ‘Bonus’ that is 10% of Salary and a column ‘Total’ = Salary + Bonus. Then rename the column ‘City’ to ‘Location’ and drop the ‘Bonus’ column.

```python
# Original df ko kharab na karne ke liye copy bana li
df6 = df.copy()

df6["Bonus"] = df6["Salary"] * 0.10
df6["Total"] = df6["Salary"] + df6["Bonus"]
print("After adding Bonus and Total:\n", df6)

df6 = df6.rename(columns={"City": "Location"})
print("After renaming City to Location:\n", df6)

df6 = df6.drop(columns=["Bonus"])
print("After dropping Bonus:\n", df6)
```

7. Create a DataFrame having some missing values (NaN). Find the count of missing values in each column, fill numeric NaNs with the column mean, and drop rows that still have missing values.

```python
data7 = {
    "Name": ["Ali", "Sara", "Ahmed", "Zara", "Usman"],
    "Age": [25, np.nan, 28, 35, np.nan],
    "City": ["Delhi", "Mumbai", None, "Pune", "Delhi"],
    "Salary": [35000, 52000, np.nan, 60000, 48000]
}

df7 = pd.DataFrame(data7)
print("Original DataFrame:\n", df7)

# Har column mein missing values ki ginti
print("Missing values in each column:\n", df7.isnull().sum())

# Numeric columns (Age, Salary) ke NaN ko unke mean se fill karna
df7["Age"] = df7["Age"].fillna(df7["Age"].mean())
df7["Salary"] = df7["Salary"].fillna(df7["Salary"].mean())
print("After filling numeric NaNs with mean:\n", df7)

# Jin rows mein ab bhi missing value hai (City) unhein drop karna
df7 = df7.dropna()
print("After dropping rows with missing values:\n", df7)
```

8. Sort the DataFrame by Salary in descending order, and then by Age in ascending order. Also print the top 3 highest-paid employees using nlargest().

```python
sorted_df = df.sort_values(by=["Salary", "Age"], ascending=[False, True])
print("Sorted DataFrame:\n", sorted_df)

top3 = df.nlargest(3, "Salary")
print("Top 3 highest-paid employees:\n", top3)
```

9. Using groupby, find the average salary per City, the number of employees per City, and the maximum salary per City. Show the result using agg() with multiple functions.

```python
print("Average salary per City:\n", df.groupby("City")["Salary"].mean())
print("Employees per City:\n", df.groupby("City")["Salary"].count())
print("Maximum salary per City:\n", df.groupby("City")["Salary"].max())

# agg() ke sath teeno ek hi table mein
result = df.groupby("City")["Salary"].agg(["mean", "count", "max"])
print("Using agg():\n", result)
```

10. Create two DataFrames (employees and departments) with a common column ‘DeptID’. Merge them using inner and left joins, then concatenate two DataFrames of the same structure using pd.concat() and remove duplicate rows.

```python
employees = pd.DataFrame({
    "EmpID": [1, 2, 3, 4],
    "Name": ["Ali", "Sara", "Ahmed", "Zara"],
    "DeptID": [101, 102, 101, 103]
})

departments = pd.DataFrame({
    "DeptID": [101, 102, 104],
    "DeptName": ["HR", "IT", "Finance"]
})

# Inner join: sirf wo rows jo dono mein common hain
inner_join = pd.merge(employees, departments, on="DeptID", how="inner")
print("Inner join:\n", inner_join)

# Left join: employees ki saari rows, match na ho to NaN
left_join = pd.merge(employees, departments, on="DeptID", how="left")
print("Left join:\n", left_join)

# Concat: do same structure wale DataFrames ko jorna
df_a = pd.DataFrame({"Name": ["Ali", "Sara"], "Age": [25, 32]})
df_b = pd.DataFrame({"Name": ["Sara", "Zara"], "Age": [32, 35]})

combined = pd.concat([df_a, df_b], ignore_index=True)
print("After concat:\n", combined)

# Duplicate rows hatana
combined = combined.drop_duplicates()
print("After removing duplicates:\n", combined)
```
