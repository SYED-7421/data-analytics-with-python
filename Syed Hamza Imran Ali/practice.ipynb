```python
# ---------- Part 1: Base Class ----------
class Person:
    def __init__(self, name, age, cnic):
        self.name = name
        self.age = age
        self.cnic = cnic

    def display_info(self):
        print("Name:", self.name)
        print("Age:", self.age)
        print("CNIC:", self.cnic)

    def role(self):
        return "Person"


# ---------- Part 2: Child Classes ----------
class Student(Person):
    def __init__(self, name, age, cnic, roll_no, marks):
        super().__init__(name, age, cnic)
        self.roll_no = roll_no
        self.marks = marks

    def average(self):
        return sum(self.marks) / len(self.marks)

    def grade(self):
        avg = self.average()
        if avg >= 80:
            return "A"
        elif avg >= 70:
            return "B"
        elif avg >= 60:
            return "C"
        else:
            return "F"

    def role(self):
        return "Student"

    def display_info(self):
        super().display_info()
        print("Roll No:", self.roll_no)
        print("Marks:", self.marks)
        print("Average:", self.average())
        print("Grade:", self.grade())


class Teacher(Person):
    def __init__(self, name, age, cnic, subject, salary):
        super().__init__(name, age, cnic)
        self.subject = subject
        self.salary = salary

    def annual_salary(self):
        return self.salary * 12

    def role(self):
        return "Teacher"

    def display_info(self):
        super().display_info()
        print("Subject:", self.subject)
        print("Monthly Salary:", self.salary)
        print("Annual Salary:", self.annual_salary())


class Admin(Person):
    def __init__(self, name, age, cnic, department, salary):
        super().__init__(name, age, cnic)
        self.department = department
        self.salary = salary

    def annual_salary(self):
        base = self.salary * 12
        bonus = base * 0.10
        return base + bonus

    def role(self):
        return "Admin"

    def display_info(self):
        super().display_info()
        print("Department:", self.department)
        print("Monthly Salary:", self.salary)
        print("Annual Salary (with bonus):", self.annual_salary())


# ---------- Part 3: Polymorphism ----------
people = []

people.append(Student("Ali", 20, "12345-1111111-1", 101, [85, 90, 78]))
people.append(Student("Sara", 21, "12345-2222222-2", 102, [55, 60, 58]))
people.append(Teacher("Ahmed", 35, "12345-3333333-3", "Math", 50000))
people.append(Teacher("Ayesha", 40, "12345-4444444-4", "English", 60000))
people.append(Admin("Bilal", 45, "12345-5555555-5", "Accounts", 45000))

for person in people:
    print("Role:", person.role())
    person.display_info()
    print("-" * 30)


# ---------- Part 4: Reports ----------
student_count = 0
teacher_count = 0
admin_count = 0

for person in people:
    if isinstance(person, Student):
        student_count += 1
    elif isinstance(person, Teacher):
        teacher_count += 1
    elif isinstance(person, Admin):
        admin_count += 1

print("Total Students:", student_count)
print("Total Teachers:", teacher_count)
print("Total Admins:", admin_count)

# Student with highest average
top_student = None
top_average = -1

for person in people:
    if isinstance(person, Student):
        avg = person.average()
        if avg > top_average:
            top_average = avg
            top_student = person

print("Top Student:", top_student.name, "with average", top_average)

# Total annual salary of staff (Teachers + Admins)
total_salary = 0

for person in people:
    if isinstance(person, Teacher) or isinstance(person, Admin):
        total_salary += person.annual_salary()

print("Total Annual Salary (Staff):", total_salary)
```

```python
jobs = soup.findAll('div',class_= 'column is-half')

jobs_data = []

for job in jobs:

    job_title = job.find('h2', class_ = 'title is-5').text.strip()

    job_subtitle = job.find('h3', class_ = 'subtitle is-6 company').text.strip()

    job_location = job.find('p', class_ = 'location').text.strip()

    job_date = job.find('p', class_ = 'is-small has-text-grey').text.strip()

    data = [job_title, job_subtitle , job_location, job_date]

    jobs_data.append(data)

print(jobs_data)




```
