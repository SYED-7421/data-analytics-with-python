```python
def calculate_grade(marks):

    if marks >= 90:

        return "A+"

    elif marks >= 80:

        return "A"

    elif marks >= 70:

        return "B"

    elif marks >= 60:

        return "C"

    else:

        return "F"

students_marks = [

    {

        "name": "Ali",

        "marks": 85

    },

    {

        "name": "Zafar",

        "marks": 92

    },

    {

        "name": "Ayan",

        "marks": 78

    },

    {

        "name": "Fayyaz",

        "marks": 65

    }

]

for student in students_marks:

    grade = calculate_grade(student["marks"])

    print(f"{student['name']} has scored {student['marks']} marks and has received grade {grade}.")
```
