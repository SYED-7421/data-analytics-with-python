```python
marks=int(input("ENTER YOUR MARKS : "))

if marks > 80:

    print("YOU GOT 'A' GRADE")

elif marks >= 60 and marks <= 80 :

    print("YOU GOT 'B' GRADE")

elif marks >= 50 and marks <= 60 :

    print("YOU GOT 'C' GRADE")

elif marks >= 45 and marks <= 50 :

    print("YOU GOT 'D' GRADE")

elif marks >= 25 and marks <= 45 :

    print("YOU GOT 'E' GRADE")

elif marks < 25 :

    print("YOU GOT 'F' GRADE")
```

```python
classes_held = int(input("ENTER THE NUMBER OF CLASSES HELD"))

classes_attended = int(input("ENTER THE NUMBER OF CLASSES ATTENDED"))

percentage = classes_attended / classes_held *100

print("PERCENTAGE OF CLASSES ATTENDED : ", percentage)

if percentage < 75 :

    medical_cause=input(" Do you have medical cause? ( 'Y' or 'N' ): ")

    if medical_cause == "Y":

        print("YOU ARE  ALLOWED TO SIT IN EXAM")
    
    else:    

        print("YOU ARE NOT ALLOWED TO SIT IN EXAM")

else:

    print("YOU ARE ALLOWED TO SIT IN EXAM")
```

```python
a = int(input("Enter the value of a:"))
b = int(input("\nEnter the value of b:"))

print("\nBefore swapping: a =", a, ", b =", b)

a = a + b
b = a - b
a = a - b

print("\nAfter swapping: a =", a)
print("b =", b)
```

```python
n = int(input("Enter number of elements: "))
list = []
for i in range(n):
    value = input(f"Enter element {i+1}: ")
    list.append(value)

print("List:", list)

for i in range(0,len(list),2):
    print("ALTERNATIVE LIST:",list[i])

```
