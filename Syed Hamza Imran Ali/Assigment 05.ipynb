```python
numbers = [12, 45, 19, 52, 42, 13, 92, 34]  # list of integers

# print all numbers
print("All numbers:")
for num in numbers:
    print(num, end=" ")
print()

# sum of all numbers
total = 0
for num in numbers:
    total += num
print("Sum of all numbers:", total)

# print only even numbers
print("Even numbers:")
for num in numbers:
    if num % 2 == 0:   # check if divisible by 2
        print(num, end=" ")
print()
```

```python
student = {
    "name": "Ali",
    "age": 20,
    "course": "Computer Science",
    "marks": [80, 65, 90]
}

# print all keys and values
for key, value in student.items():
    print(key, ":", value)

# calculate average of marks
total = 0
for mark in student["marks"]:
    total += mark
average = total / len(student["marks"])
print("Average:", average)

# check pass/fail
if average >= 50:
    print("Result: Passed")
else:
    print("Result: Failed")
```

```python
employees = [
    {"name": "Ali", "department": "IT", "salary": 50000},
    {"name": "Sara", "department": "HR", "salary": 45000},
    {"name": "Usman", "department": "Finance", "salary": 60000}
]

# print details of all employees
for emp in employees:
    print(emp["name"], "-", emp["department"], "-", emp["salary"])

# find employee with highest salary
highest = employees[0]   # assume first employee is highest to start
for emp in employees:
    if emp["salary"] > highest["salary"]:   # compare each salary
        highest = emp
print("Highest paid:", highest["name"], "-", highest["salary"])

# calculate total salary expense
total = 0
for emp in employees:
    total += emp["salary"]
print("Total salary expense:", total)
```

```python
numbers = []  # empty list to store user numbers

while True:
    num = int(input("Enter a number (-1 to stop): "))
    if num == -1:   # stop condition
        break
    numbers.append(num)

print("Complete list:", numbers)
print("Largest number:", max(numbers))
print("Smallest number:", min(numbers))
```

```python
sentence = input("Enter a sentence: ")
words = sentence.split()   # split sentence into list of words

word_count = {}   # empty dictionary to store word counts

for word in words:
    word = word.lower()   # treat "Apple" and "apple" as same word
    if word in word_count:
        word_count[word] += 1   # word seen before, increase count
    else:
        word_count[word] = 1    # first time seeing word, start at 1

print("Word Frequency:")
for word, count in word_count.items():
    print(f"{word}: {count}")
```

```python
# multiplication table from 1 to 5 using nested loops
for i in range(1, 6):        # outer loop -> controls each row (1 to 5)
    for j in range(1, 6):    # inner loop -> controls each column (1 to 5)
        print(i * j, end=" ")   # print product, stay on same line
    print()   # move to next line after inner loop finishes (new row)
```
